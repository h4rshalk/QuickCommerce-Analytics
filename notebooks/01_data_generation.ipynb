```python
import pandas as pd
import numpy as np
from faker import Faker
import random
import os

print("All libraries loaded successfully")
print(f"Pandas version: {pd.__version__}")
print(f"Numpy version: {np.__version__}")
```

```python
# Set seed for reproducibility - this ensures same data every time we run
np.random.seed(42)
random.seed(42)
fake = Faker('en_IN')  # Indian locale for realistic names and addresses

# Output path
OUTPUT_PATH = '../data/raw/'

print("Setup done. Output path:", OUTPUT_PATH)
```

```python
# ── dim_customers ──
cities_areas = {
    'Pune': ['Koregaon Park', 'Wakad', 'Baner', 'Kothrud', 'Viman Nagar', 'Hadapsar', 'Aundh'],
    'Mumbai': ['Andheri', 'Bandra', 'Powai', 'Thane', 'Borivali', 'Dadar', 'Kurla'],
    'Bangalore': ['Koramangala', 'Indiranagar', 'Whitefield', 'HSR Layout', 'BTM Layout', 'Jayanagar']
}

segments = ['New', 'Regular', 'Premium']
segment_weights = [0.3, 0.5, 0.2]  # 30% New, 50% Regular, 20% Premium

customers = []
for i in range(1, 5001):
    city = random.choice(list(cities_areas.keys()))
    area = random.choice(cities_areas[city])
    customers.append({
        'customer_id': f'CUST{i:04d}',
        'customer_name': fake.name(),
        'city': city,
        'area': area,
        'signup_date': fake.date_between(start_date='-3y', end_date='today'),
        'customer_segment': random.choices(segments, weights=segment_weights)[0]
    })

dim_customers = pd.DataFrame(customers)
dim_customers.to_csv(OUTPUT_PATH + 'dim_customers.csv', index=False)

print(f"dim_customers created: {dim_customers.shape}")
print(dim_customers.head(3))
```

```python
# ── dim_dark_stores ──
store_areas = {
    'Pune': ['Koregaon Park', 'Wakad', 'Baner', 'Kothrud', 'Viman Nagar', 
             'Hadapsar', 'Aundh', 'Magarpatta', 'Hinjewadi', 'Shivajinagar'],
    'Mumbai': ['Andheri', 'Bandra', 'Powai', 'Thane', 'Borivali', 
               'Dadar', 'Kurla', 'Malad', 'Goregaon', 'Vikhroli'],
    'Bangalore': ['Koramangala', 'Indiranagar', 'Whitefield', 'HSR Layout', 
                  'BTM Layout', 'Jayanagar', 'Marathahalli', 'Bellandur', 
                  'Electronic City', 'Yelahanka']
}

stores = []
store_counter = 1
for city, areas in store_areas.items():
    for area in areas:
        stores.append({
            'store_id': f'STR{store_counter:03d}',
            'store_name': f'{area} Dark Store',
            'city': city,
            'area': area,
            'opening_date': fake.date_between(start_date='-2y', end_date='-6m'),
            'capacity_score': random.randint(1, 10)
        })
        store_counter += 1

dim_dark_stores = pd.DataFrame(stores)
dim_dark_stores.to_csv(OUTPUT_PATH + 'dim_dark_stores.csv', index=False)

print(f"dim_dark_stores created: {dim_dark_stores.shape}")
print(dim_dark_stores.head(3))
```

```python
# ── dim_riders ──
vehicle_types = ['Bike', 'Scooter', 'Bicycle']
vehicle_weights = [0.5, 0.35, 0.15]  # 50% Bike, 35% Scooter, 15% Bicycle

riders = []
for i in range(1, 501):
    city = random.choice(list(cities_areas.keys()))
    riders.append({
        'rider_id': f'RDR{i:03d}',
        'rider_name': fake.name(),
        'joining_date': fake.date_between(start_date='-2y', end_date='today'),
        'vehicle_type': random.choices(vehicle_types, weights=vehicle_weights)[0],
        'city': city
    })

dim_riders = pd.DataFrame(riders)
dim_riders.to_csv(OUTPUT_PATH + 'dim_riders.csv', index=False)

print(f"dim_riders created: {dim_riders.shape}")
print(dim_riders.head(3))
```

```python
# ── dim_products ──
product_catalog = {
    'Dairy': ['Amul Milk 1L', 'Amul Butter 500g', 'Amul Paneer 200g', 
              'Curd 400g', 'Cheese Slices 200g'],
    'Grocery': ['Tata Salt 1kg', 'Fortune Atta 5kg', 'Saffola Oil 1L', 
                'Basmati Rice 1kg', 'Toor Dal 500g'],
    'Snacks': ['Lays Classic 50g', 'Kurkure Masala 40g', 'Oreo Biscuits 120g', 
               'Good Day Cookies 100g', 'Haldirams Bhujia 200g'],
    'Beverages': ['Coca Cola 750ml', 'Tropicana Orange 1L', 'Red Bull 250ml', 
                  'Bisleri Water 1L', 'Nescafe Coffee 50g'],
    'Personal Care': ['Dove Soap 100g', 'Colgate Toothpaste 200g', 'Head Shoulders 180ml',
                      'Dettol Handwash 200ml', 'Nivea Cream 100ml'],
    'Fruits & Vegetables': ['Banana 1kg', 'Tomato 500g', 'Onion 1kg', 
                             'Apple 4pcs', 'Spinach 250g'],
    'Frozen Foods': ['Mccain Fries 400g', 'Amul Ice Cream 500ml', 
                     'Frozen Peas 500g', 'Fish Fingers 300g', 'Chicken Nuggets 400g'],
    'Baby Care': ['Pampers Diapers S 20pcs', 'Cerelac Wheat 300g', 
                  'Johnson Baby Powder 100g', 'Mamy Poko Pants M 18pcs']
}

# Price ranges per category
price_ranges = {
    'Dairy': (25, 150),
    'Grocery': (40, 350),
    'Snacks': (20, 120),
    'Beverages': (20, 250),
    'Personal Care': (50, 400),
    'Fruits & Vegetables': (20, 100),
    'Frozen Foods': (80, 350),
    'Baby Care': (150, 800)
}

products = []
product_counter = 1
for category, items in product_catalog.items():
    for item in items:
        low, high = price_ranges[category]
        products.append({
            'product_id': f'PRD{product_counter:03d}',
            'product_name': item,
            'category': category,
            'unit_price': round(random.uniform(low, high), 2)
        })
        product_counter += 1

dim_products = pd.DataFrame(products)
dim_products.to_csv(OUTPUT_PATH + 'dim_products.csv', index=False)

print(f"dim_products created: {dim_products.shape}")
print(dim_products.head(3))
```

```python
# ── fact_orders PART 1 - Setup ──

# Pull IDs from dimension tables
customer_ids = dim_customers['customer_id'].tolist()
store_ids = dim_dark_stores['store_id'].tolist()
rider_ids = dim_riders['rider_id'].tolist()
product_ids = dim_products['product_id'].tolist()

# Match product prices
product_price_map = dim_products.set_index('product_id')['unit_price'].to_dict()

# Order settings
order_statuses = ['Delivered', 'Cancelled', 'Returned']
status_weights = [0.85, 0.10, 0.05]  # 85% delivered, 10% cancelled, 5% returned

payment_methods = ['UPI', 'Card', 'COD', 'Wallet']
payment_weights = [0.50, 0.25, 0.15, 0.10]

# Peak hours - morning and evening
morning_peak = list(range(7, 11))   # 7am to 10am
evening_peak = list(range(18, 22))  # 6pm to 9pm
normal_hours = list(range(6, 23))

print("Part 1 setup done")
print(f"Customers: {len(customer_ids)} | Stores: {len(store_ids)} | Riders: {len(rider_ids)} | Products: {len(product_ids)}")
```

```python
# ── fact_orders PART 2 - Generation ──

orders = []
num_orders = 150000

for i in range(1, num_orders + 1):
    
    # Pick hour with peak hour bias
    is_peak = random.random() < 0.45  # 45% orders during peak hours
    if is_peak:
        hour = random.choice(morning_peak + evening_peak)
    else:
        hour = random.choice(normal_hours)
    
    minute = random.randint(0, 59)
    second = random.randint(0, 59)
    order_time = f"{hour:02d}:{minute:02d}:{second:02d}"
    
    # Delivery time logic
    # During peak hours - higher chance of delay
    if hour in morning_peak or hour in evening_peak:
        actual_delivery = random.choices(
            [random.randint(8, 10),   # On time
             random.randint(11, 20),  # Slightly late
             random.randint(21, 45)], # Very late
            weights=[0.30, 0.45, 0.25]
        )[0]
        is_stockout = random.choices([0, 1], weights=[0.75, 0.25])[0]  # Higher stockout during peak
    else:
        actual_delivery = random.choices(
            [random.randint(8, 10),   # On time
             random.randint(11, 20),  # Slightly late
             random.randint(21, 45)], # Very late
            weights=[0.60, 0.30, 0.10]
        )[0]
        is_stockout = random.choices([0, 1], weights=[0.90, 0.10])[0]

    # Inject anomalies in ~2% of records
    if random.random() < 0.02:
        actual_delivery = random.choice([-5, -1, 0, 99, 120])  # Bad data

    product_id = random.choice(product_ids)
    unit_price = product_price_map[product_id]
    quantity = random.randint(1, 5)
    total_amount = round(unit_price * quantity, 2)

    order_status = random.choices(order_statuses, weights=status_weights)[0]

    # Inject ~3% duplicate orders (anomaly)
    order_id = f"ORD{i:06d}"
    if random.random() < 0.03:
        order_id = f"ORD{random.randint(1, i):06d}"  # Duplicate ID

    # Inject ~4% missing rider IDs (anomaly)
    if random.random() < 0.04:
        rider_id = None
    else:
        rider_id = random.choice(rider_ids)

    orders.append({
        'order_id': order_id,
        'customer_id': random.choice(customer_ids),
        'store_id': random.choice(store_ids),
        'rider_id': rider_id,
        'product_id': product_id,
        'order_date': fake.date_between(start_date='-1y', end_date='today'),
        'order_time': order_time,
        'quantity': quantity,
        'unit_price': unit_price,
        'total_amount': total_amount,
        'promised_delivery_mins': 10,
        'actual_delivery_mins': actual_delivery,
        'order_status': order_status,
        'is_stockout': is_stockout,
        'payment_method': random.choices(payment_methods, weights=payment_weights)[0],
        'is_late_delivery': 1 if actual_delivery > 10 else 0
    })

fact_orders = pd.DataFrame(orders)
fact_orders.to_csv(OUTPUT_PATH + 'fact_orders.csv', index=False)

print(f"fact_orders created: {fact_orders.shape}")
print(f"Total Revenue: ₹{fact_orders['total_amount'].sum():,.2f}")
print(f"Late Deliveries: {fact_orders['is_late_delivery'].sum():,}")
print(f"Stockouts: {fact_orders['is_stockout'].sum():,}")
print(f"Missing Rider IDs: {fact_orders['rider_id'].isna().sum():,}")
print(fact_orders.head(3))
```

```python
# ── Final Summary ──

tables = {
    'dim_customers': dim_customers,
    'dim_dark_stores': dim_dark_stores,
    'dim_riders': dim_riders,
    'dim_products': dim_products,
    'fact_orders': fact_orders
}

print("=" * 50)
print("PROJECT DATA GENERATION SUMMARY")
print("=" * 50)

for name, df in tables.items():
    file_size = os.path.getsize(OUTPUT_PATH + f'{name}.csv') / (1024 * 1024)
    print(f"\n{name}")
    print(f"  Rows       : {df.shape[0]:,}")
    print(f"  Columns    : {df.shape[1]}")
    print(f"  File Size  : {file_size:.2f} MB")
    print(f"  Nulls      : {df.isnull().sum().sum()}")

print("\n" + "=" * 50)
print("All CSV files saved to data/raw/")
print("=" * 50)
```
